```python
# Import libraries

import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt 
import pandas as pd 
```

Dataset is called the "Intimate Partner and Family Violence" dataset obtained from the City of Toronto's Open Data Catalogue.

https://open.toronto.ca/dataset/intimate-partner-and-family-violence/ 

```python
# define file path
file_path = "/Users/robelalemu/Documents/Projects/visualization/02_activities/assignments/Intimate Partner and Family Violence.csv"

# import csv file
df = pd.read_csv(file_path)

# prints first 5 rows of dataframe
print(df.head())
```

```python
# aggregate count by month and year
df2 = df.groupby(['REPORT_YEAR', 'FAMILY_VIOLENCE_FLAG'])['COUNT_'].count().rename('COUNT').reset_index()

# prints first 5 rows of dataframe
print(df2.head())
```

```python
#sns.set_style('ticks')

sns.set_style("whitegrid", {"font.family": "sans-serif"})

tipgraph=sns.scatterplot(data=df2,x='REPORT_YEAR',
                          y='COUNT',hue='FAMILY_VIOLENCE_FLAG',
                          palette = 'viridis'
                          )

tipgraph.set(title='Intimate Partner Violence Occurences by Year',
             xlabel = 'Year',
             ylabel = 'Counts')

tipgraph.legend_.set_title('Category')

tipgraph.get_figure().savefig('Assignment_3_python.png')
```
